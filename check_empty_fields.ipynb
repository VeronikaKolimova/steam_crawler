```python
# check empty fields
import json
from collections import Counter

def is_empty(value):
    """Проверяет, является ли значение 'пустым'."""
    if value is None:
        return True
    if isinstance(value, str) and not value.strip():
        return True
    if isinstance(value, (list, dict)) and len(value) == 0:
        return True
    return False

def main():
    # Загрузка данных
    with open('steam_top_1000.json', 'r', encoding='utf-8') as f:
        data = json.load(f)

    total_records = len(data)
    print(f" Всего записей: {total_records}")

    # Собираем все возможные ключи (на случай, если структура неоднородна)
    all_keys = set()
    for record in data:
        if isinstance(record, dict):
            all_keys.update(record.keys())

    # Считаем пустые значения по каждому полю
    empty_counts = Counter()
    missing_counts = Counter()  # поля, которых вообще нет в записи

    for record in data:
        if not isinstance(record, dict):
            continue
        for key in all_keys:
            if key not in record:
                missing_counts[key] += 1
            elif is_empty(record[key]):
                empty_counts[key] += 1

    # Выводим результат
    print("\nОтчёт о пустых или отсутствующих полях:")
    print("-" * 50)

    any_issue = False
    for key in sorted(all_keys):
        missing = missing_counts[key]
        empty = empty_counts[key]
        total_empty = missing + empty

        if total_empty > 0:
            any_issue = True
            pct = (total_empty / total_records) * 100
            print(f"Поле '{key}':")
            if missing > 0:
                print(f"  — отсутствует в {missing} записях")
            if empty > 0:
                print(f"  — пустое в {empty} записях")
            print(f"  → всего: {total_empty} ({pct:.1f}%)")
            print()

    if not any_issue:
        print("Все поля заполнены. Пустых или пропущенных значений не обнаружено.")

if __name__ == "__main__":
    main()
```
